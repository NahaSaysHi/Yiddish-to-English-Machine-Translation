1.

Source sentence:

> נאָך יאָרן פון שרייבן, דענקן, קרענקן, האָט עס ענדלעך פּאַסירט. דער בחור וועט אי"ה מקדש זיין בשעה טובה ומוצלחת און אין אַ מזלדיקע שעה וועט מען שטעלן חופה פאַר אים און זיין בת זוג וואָס מען האָט אים באַשערט זייט פערציק טעג איידער ער איז באַשאַפן געוואָרן.

Deep L translation:


> After years of writing, thinking, and suffering, it finally happened. God willing, the groom will be sanctified at a good and auspicious time, and at a fortunate hour a chuppah will be set up for him and his soulmate, who was destined for him forty days before he was created.

Analysis:

This is a very good translation! I was impressed that אי"ה was recognized and correctly tranlated to 'God willing'. I also thought it was interesting that חופה was tranlated to its transliteration 'chuppah' rather than translated to 'cannopy'. Many Jewish speakers will use the word 'chuppah' in their English dialect, so I think the translation makes sense and for many translation contexts would feel more appropriate!



2.

Source sentence:


> וועט איר מיר פרעגן, און אַז דאָס קינד איז נאָך נישט באַשאַפן ווי איז שייך עס משדך צו זיין?.


DeepL translation:



> Will you ask me, and that the child has not yet been created as is fitting for it to be matched?

My translation:


> You will ask me, since the child has not yet been created, how is it appropriate for it to be matched?


Analysis:

DeepL's translation just doesn't make a whole lot of sense in English. The author is not asking to be asked ('will you ask me') but rather presuming that he will be asked ('you will ask me'). Also, ווי איז can mean 'as is' but it can also mean 'how is'. In this context, the ווי איז means 'how is (it)'. The phrases, if seperated, are translated correctly, but they aren't translated correctly in context and the result is a sentence that doesn't make a whole lot of sense.

3.

Source sentence:


> וועל איך אייך ענטפערן אַז איר קענט נישט אנזערע שדכנים.

DeepL translation:


> I'll answer you that you can't contact our matchmakers.

My translation:



> I'll answer that you don't know our matchmakers.

Analysis:

I would describe this translation as incorrect! איר קענט נישט can mean "you cannot" but it can also mean "you don't know". In this context, it means the latter. While I can forgive it for not understanding the context, I don't understand where the word 'contact' comes from. Given that it had to add a word that wasn't even in the origal sentence to make the "you cannot" translatin make sense, I don't know why it wouldn't choose the other translation.

Another thing that is interesting is the translation to "I'll answer you that" instead of "I'll answer that". While the Yiddish does literally say "I'll answer you that", it makes more sense in Standard American English to say "I'll answer that". That being said, some native Yiddish speakers would say "I'll answer you that" in English. But if I were a native Yiddish speaker using Google Translate to translate Yiddish text, I would probably want the translator to give me a SAE translation.

Source sentence:

DeepL translation:

>If you've been following my series on The Man Who Sanctifies, you'll already know that a child is born to us; God, in His great mercy, has granted us a son, He was circumcised, weaned, and had his brit milah; then he was taken to the study hall, made his bar mitzvah, and sent to yeshiva, And when, thank God, he reached the age, the community began to wish him well, matchmakers called, and we began to ‘listen’ (or in English, listening), They proposed ‘things,’ some of which we didn't want, and others they didn't want, He was asked again; he went out with her once, and again, and once more, and each time came home later. He didn't recount everything anymore, He wandered around the house without speaking, just tugging at his beard, and on a beautiful day—or rather, evening—the plates were broken, Everyone shouted “Mazel Tov,” and we became in-laws.

Analysis:

Overall, especially for the setence's length, this is a preety good translation. That being said, the translation "He was circumcised, weaned, and had his brit milah" is repetitive since being circumcised and having a bris milah is the same exact thing. While this indeed repeated in the original sentence, the repetition comes from including a quote from the Torah or the Talmud. It also skips that he had an upshern, which is the practice of cutting a three year-old boy's hair. I would change the translation to "He was circumcised and had his hair cut at age three" or "He had a bris milah and an upshernish". I would change the translation "then he was taken to the study hall" to "then he was sent to school" since study hall implies an older age student than the original word חדר does. I would also change "made his bar mitzvah" to "given a bar mitzvah" because "then he was taken to the study hall, made his bar mitzvah, and sent to yeshiva" makes less sense than "then he was taken to the study hall, given a bar mitzvah, and sent to yeshiva".

Source sentence:

> שבת גריסט מען אייך, גוט שבת ר' מחותן!

DeepL translation:

> Shabbat greetings to you, Good Shabbat, Rabbi Chatan!

My translation:

> On Shabbos, they greet you, "Good Shabbos Reb father-in-law"!

Analysis:

'Chatan' means 'groom'. On the other hand, 'm'chutan' means father-in-law. Therefore, translating מחותן (m'chutan / father-in-law) to 'Chatan' (groom) is incorrect! Also, the translation makes it sound like he is giving Shabbos greetings when he is really saying that "on Shabbos they greet you".

Source sentence:

> כלומר ווי דער וועלט איז צוטיילט אויף צווייען: די וואָס האָבן נאָך נישט חתונה געמאַכט און די וואָס יאָ.

DeepL translation:

> That is, how the world is divided into two: those who haven't gotten married yet and those who have.

My translation:

> As if to say that the world is divided into two: Those who haven't married off yet and those who have.

Analysis:

The DeepL translation doesn't fully communicate that the author is saying that the reaction from other people (described in the prior sentences) is what gives the message that the world is divided into two. Also, the DeepL translation makes it seem like the division is between being single and getting married, but
חתונה געמאַכט ("made a wedding") communciates being the parent of the person getting married and context (the next sentence talks about this division being about getting old and growing a grey beard) confirms this.

Overall analyis:

Acutal errors often came from subtle differences. For example, having a wedding (getting married) vs. making a wedding (marrying off) or being a chosson (groom) vs. m'chutan (father-in-law). Difference in phrase structure also caused poor translation. For example, what literally translates based on phrase order to "will you ask me" actually means "you will ask me". Finally, many errors were caused by words / phrases with multiple correct translations but only some correct in context.

```python
# IBM Model 1 matched with pseudocode
# Variables: t(e|f), count(e|f), total(f), s_total(e)

from collections import defaultdict, Counter

def training_IBM_Model_1(tokenized_f, tokenized_e, num_iters=5, add_null=True):
    # English vocabulary = e_vocab
    e_vocab_set = set()
    i = 0
    while i < len(tokenized_e):
        e_sent = tokenized_e[i]
        j = 0
        while j < len(e_sent):
            e_vocab_set.add(e_sent[j])
            j += 1
        i += 1
    e_vocab = sorted(list(e_vocab_set))

    # Yiddish vocabulary = f_vocab
    f_vocab_set = set()
    i = 0
    while i < len(tokenized_f):
        f_sent = tokenized_f[i]
        j = 0
        while j < len(f_sent):
            f_vocab_set.add(f_sent[j])
            j += 1
        i += 1
    if add_null:
        f_vocab_set.add("<NULL>")
    f_vocab = list(f_vocab_set)


    t = {}
    if len(e_vocab) == 0:
        raise ValueError("No English tokens.")
    uniform_prob = 1.0 / float(len(e_vocab))

    i = 0
    while i < len(f_vocab):
        f = f_vocab[i]
        t[f] = {}
        j = 0
        while j < len(e_vocab):
            e = e_vocab[j]
            t[f][e] = uniform_prob
            j += 1
        i += 1


    it = 1
    while it <= num_iters:
        # count(e|f)=0 and total(f)=0
        count = defaultdict(lambda: defaultdict(float))  # count[f][e]
        total = defaultdict(float)                       # total[f]


        s = 0
        while s < len(tokenized_f) and s < len(tokenized_e):
            f_sent = tokenized_f[s]
            e_sent = tokenized_e[s]


            f_words = []
            if add_null:
                f_words.append("<NULL>")
            j = 0
            while j < len(f_sent):
                f_words.append(f_sent[j])
                j += 1


            s_total = defaultdict(float)  # s_total[e]
            j = 0
            while j < len(e_sent):
                e = e_sent[j]
                s_total_e = 0.0
                k = 0
                while k < len(f_words):
                    f = f_words[k]
                    s_total_e += t[f].get(e, 0.0)
                    k += 1
                s_total[e] = s_total_e
                j += 1


            j = 0
            while j < len(e_sent):
                e = e_sent[j]
                denom = s_total[e]
                if denom != 0.0:
                    k = 0
                    while k < len(f_words):
                        f = f_words[k]
                        frac = t[f].get(e, 0.0) / denom
                        count[f][e] += frac
                        total[f] += frac
                        k += 1
                j += 1

            s += 1

        # t(e|f) = count(e|f)/total(f)
        i = 0
        while i < len(f_vocab):
            f = f_vocab[i]
            Z = total[f]
            if Z != 0.0:
                j = 0
                while j < len(e_vocab):
                    e = e_vocab[j]
                    t[f][e] = count[f][e] / Z
                    j += 1
            else:

                j = 0
                while j < len(e_vocab):
                    e = e_vocab[j]
                    t[f][e] = uniform_prob
                    j += 1
            i += 1

        print(f"Finished EM iteration {it}/{num_iters}")
        it += 1

    return t


```

```python
def print_blocks(t, tokenized_f, top_f=10, top_e=5, ignore=None, decimals=3):
    if ignore is None:
        ignore = set()


    freq_f = Counter(w for sent in tokenized_f for w in sent if w not in ignore and w.strip() != "")
    most_common_f = [w for w, _ in freq_f.most_common(top_f)]

    for f in most_common_f:

        best_all = sorted(t.get(f, {}).items(), key=lambda x: x[1], reverse=True)


        best = []
        i = 0
        while i < len(best_all) and len(best) < top_e:
            e, p = best_all[i]
            if e not in ignore and e.strip() != "":
                best.append((e, p))
            i += 1

        print("\n" + "=" * 60)
        print(f"f (source): {f}    count: {freq_f[f]}")
        print("-" * 60)
        i = 0
        while i < len(best):
            e, p = best[i]
            print(f"{i+1}. {e:<10}  {p:.{decimals}f}")
            i += 1




```

```python
n = min(len(tokenized_yid), len(tokenized_eng))
tokenized_yid = tokenized_yid[:n]
tokenized_eng = tokenized_eng[:n]


N = min(10_000, n)
train_f = tokenized_yid[:N] #Yiddish tokenized sentences
print(train_f)
train_e = tokenized_eng[:N] #English tokenized sentences
print(train_e)
t_ef = training_IBM_Model_1(train_f, train_e, num_iters=5, add_null=True) #t_ef = translation table

# Optional punctuation filter
punct = {".", ",", "?", "!", ":", ";", '"', "'", "—", "-", "(", ")", "[", "]"}
print_blocks(t_ef, train_f, top_f=10, top_e=5, ignore=punct)
```

```python
import math, random

# Finds the value of log p of english given foreign for IBM Model 1
def log2_p_e_given_f_ibm1(f_sent, e_sent, t, add_null=True, null_token="<NULL>", smooth=1e-12):
    f_words = ([null_token] + f_sent) if add_null else list(f_sent)
    m = len(f_sent)
    denom = (m + 1) if add_null else m
    if denom <= 0:
        return float("-inf")

    log2p = 0.0
    for e in e_sent:
        s = 0.0
        for f in f_words:
            s += t.get(f, {}).get(e, 0.0)
        s = max(s, smooth)
        log2p += math.log2(s / denom)
    return log2p
```

```python
# Finds the perplexity value of the sentence.
#I have used the formula that was suggested in class directly and did not normalize the value.
def perplexity_single(f_sent, e_sent, t, **kwargs):

    log2p = log2_p_e_given_f_ibm1(f_sent, e_sent, t, **kwargs)
    return 2 ** (-log2p)
```

```python
def build_english_vocab(tokenized_e, ignore=None):
    if ignore is None:
        ignore = set()
    vocab = []
    for sent in tokenized_e:
        for w in sent:
            if w not in ignore and w.strip() != "":
                vocab.append(w)
    return vocab
```

```python
# I have generated random english sentences to verify that perplexity value is higher for these as compared to translated sentences.
def random_english_sentence(vocab, length):
    return [random.choice(vocab) for _ in range(length)]
```

```python
# This method compares the perplexity value of random sentences to translated sentences.
def compare_real_vs_random(train_f, train_e, t_ef, vocab, trials=5, seed=42,
                                   add_null=True, null_token="<NULL>"):
    random.seed(seed)

    for trial in range(trials):
        idx = random.randrange(len(train_f))
        f_sent = train_f[idx]
        e_real = train_e[idx]
        e_rand = random_english_sentence(vocab, len(e_real))

        pp_real = perplexity_single(f_sent, e_real, t_ef, add_null=add_null, null_token=null_token)
        pp_rand = perplexity_single(f_sent, e_rand, t_ef, add_null=add_null, null_token=null_token)

        print("\n" + "=" * 90)
        print(f"Trial {trial+1} (idx={idx})")
        print("Yiddish f:", " ".join(f_sent))
        print("Real EN :", " ".join(e_real))
        print("Rand EN :", " ".join(e_rand))
        print(f"PP(real) = {pp_real:.4f}")
        print(f"PP(rand) = {pp_rand:.4f}")
```

```python
english_vocab = build_english_vocab(train_e, ignore=punct)

compare_real_vs_random(
    train_f=train_f,
    train_e=train_e,
    t_ef=t_ef,
    vocab=english_vocab,
    trials=5,
    seed=42,
    add_null=True,
    null_token="<NULL>"
)
```
